## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*



**Question:** Which pages should a reviewer look at first when they only have time
for a handful?

**Decision this improves:** where limited editor/reviewer time goes each week.

**Who acts on it:** a content strategist or SEO editor, working down a ranked list ,
deciding to refresh, expand, protect, prune, or just keep monitoring each page.

**Cost of getting it wrong:**
- Flag a fine page → wasted reviewer time, not a big deal but adds up.
- Miss a page that's actually declining → the problem keeps getting worse before
  anyone notices, which is the more expensive mistake.
- Since reviewers can only realistically get through maybe 20-50 pages a week, what
  matters most is getting the *top* of the list right, not the whole list.

**Why ML instead of just a rule:** A hand rule like "stale AND visible" only checks
one interaction. What I saw in notebook 02 is that a model , even a fully readable
one — picks up on combinations a simple if/else misses (impressions, position,
freshness, content age, CTR all interacting). That's not a small gap either
Precision@50 nearly tripled over the rule-based baseline in the reference pipeline.

**One thing I already know to watch out for:** the label the starter uses
(`trend_direction == "down"`) is calculated from the *current* window, not a real
future outcome — so it's a proxy, not the real target. Once I have warehouse access
I want to rebuild this as an actual forward-looking label (past 90 days predicting
the next 30), like the lane guide recommends.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# how many pages actually have enough traffic to be worth looking at
visible = df[df["impressions_90d"] >= 100]
print(f"Pages with >=100 impressions_90d: {len(visible)} of {len(df)} ({len(visible)/len(df):.1%})")

# what fraction of those are labeled "declining"
decline_rate = (visible["trend_direction"] == "down").mean()
print(f"Decline rate among visible pages: {decline_rate:.1%}")

# how many are both stale AND visible — the exact hand-rule I tried in notebook 02
stale_visible = visible[visible["days_since_last_update"] >= 180]
print(f"Stale AND visible pages: {len(stale_visible)} ({len(stale_visible)/len(visible):.1%} of visible)")
```
